```python
import numpy as np

# Máscara de filas con prima == 800
mask_800 = df["Prima"] == 800

# --- Modelo A: quitar TODAS las filas con prima = 800 ---
df_A = df[~mask_800].copy()

# --- Modelo B: quitar el 60% aleatorio de las filas con prima = 800 ---
idx_800 = df[mask_800].index
idx_borrar = np.random.RandomState(42).choice(
    idx_800,
    size=int(round(0.6 * len(idx_800))),
    replace=False
)
df_B = df.drop(index=idx_borrar).copy()

# Comprobación
print(f"Original:  {len(df)} filas | prima=800: {mask_800.sum()}")
print(f"Modelo A:  {len(df_A)} filas | prima=800: {(df_A['Prima'] == 800).sum()}")
print(f"Modelo B:  {len(df_B)} filas | prima=800: {(df_B['Prima'] == 800).sum()}")
```

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

SEED = 42
df = df.reset_index(drop=True)
train, test = train_test_split(df, test_size=0.2, random_state=SEED)

idx_800 = train[train["Prima"] == 800].index
idx_borrar = np.random.RandomState(SEED).choice(idx_800, size=int(round(0.6 * len(idx_800))), replace=False)
train_B = train.drop(index=idx_borrar)

cols_drop = ["ID", "Proposito"]
X_train_B = train_B.drop(columns=["Prima"] + cols_drop)
y_train_B = train_B["Prima"]
X_test = test.drop(columns=["Prima"] + cols_drop)
y_test = test["Prima"]

model_B = RandomForestRegressor(n_estimators=200, random_state=SEED, n_jobs=-1)
model_B.fit(X_train_B, y_train_B)

def evaluar(model, X, y, nombre):
    pred = model.predict(X)
    print(nombre, "MAE:", round(mean_absolute_error(y, pred), 2), "RMSE:", round(np.sqrt(mean_squared_error(y, pred)), 2), "R2:", round(r2_score(y, pred), 3))

mask = test["Prima"] != 800
evaluar(model_B, X_test, y_test, "Test completo")
evaluar(model_B, X_test[mask], y_test[mask], "Test sin 800")
evaluar(model_B, X_test[~mask], y_test[~mask], "Solo prima=800")
```

División de los datos en train y test

```python
### Hacemos train/test con las variables más correlacionadas según el método Pearson
```

```python
from sklearn.model_selection import train_test_split

X = df[['Ratio_Deuda_Ingresos', 'Monto_Inicial', 'Duracion', 'Ingresos', 'prob_pago']] 
y = df[["Prima"]]

X_train, X_test, y_train, y_test = train_test_split(
                                        X, y,
                                        train_size = 0.8,
                                        random_state = 42,
                                        shuffle = True)
```

Estantarizamos los datos del train

```python
from sklearn.preprocessing import StandardScaler

# Crear el estandarizador
scaler = StandardScaler()

# Aprender la media y desviación típica SOLO del entrenamiento
X_train_scaled = scaler.fit_transform(X_train)

# Aplicar esa misma transformación al conjunto de prueba
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression

modelo_lineal = LinearRegression()

modelo_lineal.fit(X_train_scaled, y_train)

y_pred_lineal = modelo_lineal.predict(X_test_scaled)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Calcular métricas
mae = mean_absolute_error(y_test, y_pred_lineal)
mse = mean_squared_error(y_test, y_pred_lineal)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred_lineal)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R²:", r2)
```
